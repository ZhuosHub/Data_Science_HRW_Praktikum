# Aufgabe 2 · Fehlende Werte behandeln: erst die Art, dann der Weg

**📖 Erklärung.** Aus der Vorlesung: Erst bestimmen Sie die **Art** der Lücke, mit drei Fragen in dieser Reihenfolge:

1. Lässt sich der fehlende Wert aus anderen Spalten oder Zeilen genau bestimmen? → **MD** (missing by design)
2. Gibt es einen guten Grund, warum das Fehlen vom Wert selbst abhängt? → **NMAR**
3. Sagen andere Spalten etwas darüber, ob ein Wert fehlt? → **MAR** · sonst **MCAR**

Dann wählen Sie den **Weg**:

| Art | Weg | pandas |
|---|---|---|
| MD | nichts schätzen – ablesen, eine Markierungsspalte macht den Grund sichtbar | – |
| NMAR | aus den Daten nicht zu reparieren: Grund klären, offen berichten | – |
| MAR | je Gruppe mit dem Mittelwert füllen | `groupby(…)[…].transform('mean')` · `fillna` |
| MCAR | Zeilen löschen oder mit dem Mittelwert füllen | `dropna` · `fillna(s.mean())` |

Und immer aufschreiben: welcher Weg und warum (Aufgabe 8).

**Im Projekt:** Karte 2 `pdays = -1` heißt „nie kontaktiert“ – eine Markierungsspalte wie in 2.1. Karte 4 fehlende
Preise, Karte 6 leere Spalten bei `payment_type` 0: behalten oder löschen – so wie in 2.1 und 2.2 begründet.

```python
# ===== Vorgemacht (2.1): MD – das Lieferdatum =====
ohne_datum = orders[orders['order_delivered_customer_date'].isna()]      # Bestellungen ohne Lieferdatum
print(ohne_datum['order_status'].value_counts())                          # warum fehlt es?
orders['nicht_geliefert'] = orders['order_status'] != 'delivered'         # Markierungsspalte: True = nicht geliefert
print('nicht geliefert:', orders['nicht_geliefert'].sum())
```

Fast alle Bestellungen ohne Lieferdatum sind **nicht geliefert** (`shipped`, `canceled`, `unavailable` …): Ein
Lieferdatum **gibt es nicht** – das sagt die Spalte `order_status`. Das ist **MD**: nichts füllen. Die neue Spalte
`nicht_geliefert` macht den Grund sichtbar. Nur 8 Bestellungen sind `delivered` und haben trotzdem kein Datum – das
sind echte Lücken; bei 96 478 gelieferten Bestellungen fallen sie nicht ins Gewicht.

```python
# ===== Vorgemacht (2.2): MCAR – zwei Produkte ohne Maße =====
print(produkte[produkte['product_length_cm'].isna()])                    # 2 von 32 951 Produkten, kein Muster
produkte = produkte.dropna(subset=['product_length_cm'])                  # löschen: nur Zeilen ohne Länge fallen weg
print('Produkte jetzt:', len(produkte))
```

Zwei Zeilen, kein erkennbarer Grund, kein Zusammenhang mit anderen Spalten: **MCAR**. Löschen ist hier in Ordnung –
der Mittelwert der übrigen 32 949 Produkte ändert sich dadurch kaum.

## 2.3 · Jetzt Sie: das Gewicht je Kategorie füllen

Jetzt fehlen noch die 4 Gewichte, die in 1.2 aus 0 g zu `NaN` wurden. Löschen wollen wir die Produkte nicht – ihre
Maße und ihre Kategorie stimmen ja. Wir **füllen** das Gewicht, und zwar mit dem Mittelwert **ihrer Kategorie**:
Bettwäsche wiegt anders als Möbel, also ist der Mittelwert der eigenen Gruppe die bessere Schätzung – so füllt man
bei MAR.

`groupby('product_category_name')['product_weight_g'].transform('mean')` schreibt **in jede Zeile** den Mittelwert
ihrer Kategorie – eine Spalte, genauso lang wie die Tabelle. `fillna(…)` setzt diesen Wert nur dort ein, wo `NaN` steht.

```python
# ===== Jetzt Sie (2.3): je Kategorie füllen =====
mittel_je_kategorie = produkte.groupby('product_category_name')['product_weight_g'].transform('mean')   # fertig
gewicht_gefuellt = None     # TODO: produkte['product_weight_g'], Lücken mit mittel_je_kategorie füllen – .fillna(…)

# --- Check ---
if gewicht_gefuellt is None:
    print('… noch offen')
elif gewicht_gefuellt.isna().sum() != 0:
    print(f'✗ prüfen: noch {gewicht_gefuellt.isna().sum()} Lücken – produkte[\'product_weight_g\'].fillna(mittel_je_kategorie)')
elif len(gewicht_gefuellt) != len(produkte):
    print('✗ prüfen: die Spalte muss so lang sein wie produkte – fillna, nicht dropna')
else:
    neu = gewicht_gefuellt[produkte['product_weight_g'].isna()]
    produkte['product_weight_g'] = gewicht_gefuellt          # übernommen
    wert = f'{neu.iloc[0]:,.1f}'.replace(',', ' ').replace('.', ',')
    print(f'✓ richtig · 4 Lücken gefüllt, je mit {wert} g – dem Mittelwert von cama_mesa_banho '
          '(Bettwäsche, Tisch, Bad). In produkte übernommen.')
```

# Aufgabe 5 · Datentyp: Text, Datum, Zahl

**📖 Erklärung.** Der gespeicherte Typ muss zur Bedeutung passen. Drei typische Fälle:

- **Datum als Text:** `'2017-10-02 10:56:33'` sieht aus wie ein Datum, ist aber Text. Mit Text kann man nicht rechnen
  (keine Lieferdauer). `pd.to_datetime(…)` macht ein Datum daraus; danach liefert `.dt` Jahr, Wochentag, Stunde.
- **Nummer als Zahl:** Eine Postleitzahl ist ein Name, keine Zahl. Als Zahl gespeichert, verliert `01151` die führende
  Null. `astype(str).str.zfill(5)` füllt vorne mit Nullen auf 5 Stellen auf.
- **Zahl als Text:** kennen Sie aus der Vorlesung (`str.replace` und `astype(int)`), ⭐ 9.2 zeigt es an Preisen.

Nach jeder Umwandlung: `dtypes` ansehen und prüfen, ob noch alle Zeilen da sind.
