## Spickzettel: die Formeln der Vorlesung in Python

| Vorlesung | Python |
|---|---|
| 0,5 hoch 6 | `0.5 ** 6` – Potenz mit `**`, Dezimalpunkt statt Komma |
| Wurzel √400 | `np.sqrt(400)` ergibt 20 |
| Standardabweichung s · Standardfehler SE = s / √n | `spalte.std()` · `spalte.sem()` oder `s / np.sqrt(n)` |
| je Gruppe: Anzahl, Mittelwert, s, SE | `tabelle.groupby('gruppe')['spalte'].agg(['size', 'mean', 'std', 'sem'])` |
| einen Wert aus dieser Tabelle holen | `m.loc['mit', 'mean']` – erst die Zeile, dann die Spalte |
| SE_Diff = √(SE_A² + SE_B²) | `np.sqrt(se_a**2 + se_b**2)` |
| z = Unterschied / SE_Diff | `diff / se_diff` |
| p-Wert: Fläche der Glocke außerhalb ± z | `2 * (1 - stats.norm.cdf(z))` |
| 95-%-Konfidenzintervall | `diff - 2 * se_diff` bis `diff + 2 * se_diff` |
| Gegenprobe: der t-Test | `stats.ttest_ind(a, b, equal_var=False)` |

# Teil 2 · Wie stark wackelt ein Mittelwert? (ab 16:55, selbst, ca. 8 Minuten)

Aus der Vorlesung: Wie groß sind die Studierenden im Mittel? Den wahren Wert kennt in Wirklichkeit niemand – „nur Gott“.
**Im Computer spielen wir Gott:** Wir erzeugen 10 000 Studierende und kennen deshalb den wahren Wert, 172 cm. Dann tun
wir, was man in Wirklichkeit tut: 100 Personen zufällig auswählen und messen. Die Startwerte sind so gewählt, dass
genau die drei Stichproben von der Folie herauskommen: 172,8 · 171,5 · 170,6 cm.

> 📘 Hängen Sie an der Schreibweise? Ein Klick öffnet das Grundlagen-Heft aus Woche 2 in einem neuen Tab an der richtigen Stelle: [Abschnitt 4 · Funktionen aufrufen](https://colab.research.google.com/github/ZhuosHub/Data_Science_HRW_Praktikum/blob/main/W2/W2_Praktikum_Grundlagen.ipynb#scrollTo=abschnitt-4) und [Abschnitt 5 · pandas: Spalten und Tabellen](https://colab.research.google.com/github/ZhuosHub/Data_Science_HRW_Praktikum/blob/main/W2/W2_Praktikum_Grundlagen.ipynb#scrollTo=abschnitt-5).

```python
# ===== Vorgemacht (2.1): 10 000 Studierende – und drei Stichproben zu je 100 =====
rng = np.random.default_rng(172)
alle = pd.Series(rng.normal(172, 10, size=10_000))     # 10 000 Körpergrößen: Mitte 172 cm, s etwa 10 cm
alle = (alle - alle.mean() + 172).round(1)            # auf genau 172 cm im Mittel geschoben
wahr = alle.mean()                                     # der wahre Wert – den kennt sonst nur „Gott“
print(f'wahrer Wert (alle 10 000): {wahr:.1f} cm')

for nr, startwert in enumerate((631, 14, 163), start=1):   # dreimal 100 Personen zufällig auswählen
    print(f'Stichprobe {nr}: Mittelwert {alle.sample(100, random_state=startwert).mean():.1f} cm')

stichprobe = alle.sample(100, random_state=631)      # ab jetzt rechnen wir mit der ersten Stichprobe
mittel = stichprobe.mean()                             # ihr Mittelwert
s = stichprobe.std()                                   # ihre Standardabweichung (wie in Woche 2, durch n − 1)
n = len(stichprobe)                                    # ihre Größe: 100
print(f'erste Stichprobe: n = {n}, Mittelwert {mittel:.1f} cm, s = {s:.1f} cm')
```

Der Mittelwert wackelt: Jede Stichprobe gibt eine andere Zahl. Wie stark er wackelt, sagt der **Standardfehler**
SE = s / √n. Und das **Netz** ist Mittelwert ± 2 · SE.

## 2.2 · Jetzt Sie: Standardfehler und Netz der ersten Stichprobe

```python
# ===== Jetzt Sie (2.2): Standardfehler und Netz =====
se = None          # TODO: Standardfehler SE = s / √n (s und n aus 2.1)
unten = None       # TODO: Mittelwert minus 2 Standardfehler
oben = None        # TODO: Mittelwert plus 2 Standardfehler

# --- Check ---
if se is None or unten is None or oben is None:
    print('… noch offen')
elif abs(se - s / np.sqrt(n)) > 1e-9:
    print(f'✗ prüfen: erwartet SE = s / √n = {s / np.sqrt(n):.2f} cm – die Wurzel ist np.sqrt(n)')
elif abs(unten - (mittel - 2 * se)) > 1e-9 or abs(oben - (mittel + 2 * se)) > 1e-9:
    print('✗ prüfen: das Netz ist Mittelwert ± 2 · SE')
else:
    print(f'✓ richtig · SE = {se:.1f} cm · Netz {unten:.1f} bis {oben:.1f} cm – wie auf der Folie. '
          f'Fängt es den wahren Wert {wahr:.0f}? ' + ('Ja.' if unten <= wahr <= oben else 'Nein.'))
```

## 2.3 · Die Glocke der Mittelwerte (Folie „Warum ± 2?“)

**🔮 Vorhersage:** Der Computer zieht gleich 1 000 Stichproben zu je 100 Personen. Wie breit streuen die
1 000 Mittelwerte – eher 10 cm wie einzelne Personen oder eher 1 cm? Doppelklick, Vermutung eintragen, dann ausführen.

*Meine Vorhersage:* …

```python
# ===== Vorgemacht (2.3): 1 000 Stichproben – die Glocke der Mittelwerte =====
mittelwerte = pd.Series([alle.sample(100, random_state=i).mean() for i in range(1000)])   # 1 000 Mittelwerte
fig, (links, rechts) = plt.subplots(1, 2, figsize=(11, 3.5), sharex=True)               # sharex: gleiche cm-Achse
links.hist(alle, bins=40, color='#999999')
links.set_title('10 000 einzelne Studierende: s ≈ 10 cm')
rechts.hist(mittelwerte, bins=30, color='#4F8F88')
rechts.axvline(wahr, color='black', linestyle='--')                                     # der wahre Wert
rechts.set_title('1 000 Mittelwerte aus je 100: SE ≈ 1 cm')
for achse in (links, rechts):
    achse.set_xlabel('Körpergröße (cm)')
plt.show()
print(f'Streuung der 1 000 Mittelwerte: {mittelwerte.std():.2f} cm · Formel s / √n: {s / np.sqrt(n):.2f} cm')
print(f'Anteil der Mittelwerte zwischen 170 und 174 cm (± 2 SE um 172): {mittelwerte.between(170, 174).mean():.1%}')
```

Wie auf der Folie: Auch Mittelwerte bilden eine Glocke um den wahren Wert (Zentraler Grenzwertsatz), aber zehnmal
schmaler als die einzelnen Personen – ihre Breite ist der Standardfehler. 95 % der Mittelwerte liegen höchstens
2 SE vom wahren Wert entfernt. Umgekehrt gelesen: In 95 % der Fälle liegt der wahre Wert höchstens 2 SE von unserem
Mittelwert entfernt. Deshalb fängt das Netz Mittelwert ± 2 · SE den wahren Wert in 95 % der Fälle.

## 3.2 · Woher kommt s? Die Standardabweichung in vier Schritten

Auf der Folie stand nur das Ergebnis, s ≈ 50 mit und s ≈ 28 ohne Taylor. Hier sehen Sie, wie es aus den Zahlen
entsteht – die Formel s = √( Σ (x − x̄)² / (n − 1) ) in vier Schritten: Abstand zum Ø · quadrieren und addieren ·
durch n − 1 teilen · Wurzel ziehen.

```python
# ===== Vorgemacht (3.2): s ohne Taylor in vier Schritten =====
abweichung = ohne - ohne.mean()             # Schritt 1: Abstand zum Ø – für jedes Spiel auf einmal
quadratsumme = (abweichung ** 2).sum()      # Schritt 2: quadrieren und addieren
varianz = quadratsumme / (len(ohne) - 1)    # Schritt 3: durch n − 1 teilen
s_ohne = np.sqrt(varianz)                   # Schritt 4: Wurzel ziehen
print('Abweichungen:', list(abweichung), '· Quadratsumme', quadratsumme, '· Varianz', varianz, '· s =', round(s_ohne, 1))
print('mit der Funktion: ohne.std() =', round(ohne.std(), 1))     # .std() teilt ebenfalls durch n − 1
```

## Kür C · Woher kommen die 12 €? (Frage 3, Folie „Liegt es an der App?“)

Die 800 Kunden aus dem Auftrag als Tabelle – ausgedacht, aber mit genau den Zahlen der Folien. Erst der Vergleich
App gegen Web wie in Frage 1, dann **Gleiches mit Gleichem**: getrennt nach Stammkunden und Neukunden. Das ist die
Zeile für Frage 3 von der Folie „Vier Fragen – vier Werkzeuge“. In 4.2 stand die Tabelle schon fertig da – hier
entsteht sie aus den 800 einzelnen Kunden.

```python
# ===== Vorgemacht (Kür C): die 800 Kunden aus dem Auftrag (ausgedacht, mit den Zahlen der Folien) =====
rng = np.random.default_rng(28)


def kunden_erzeugen(typ, gruppe, anzahl, mittel, s):
    """anzahl Kunden mit genau diesem Mittelwert und dieser Standardabweichung, schief verteilt wie echte Umsätze."""
    x = rng.gamma(0.8 * (mittel / s) ** 2, size=anzahl)      # schief: viele kleine, wenige große Werte, keiner unter 0
    x = mittel + s * (x - x.mean()) / x.std(ddof=1)          # auf den gewünschten Mittelwert und s gebracht
    return pd.DataFrame({'typ': typ, 'gruppe': gruppe, 'gewinn': x})


kunden = pd.concat([kunden_erzeugen('Stammkunde', 'App', 300, 67, 59.444666),
                    kunden_erzeugen('Neukunde', 'App', 100, 47, 59.444666),
                    kunden_erzeugen('Stammkunde', 'Web', 100, 65, 79.628535),
                    kunden_erzeugen('Neukunde', 'Web', 300, 45, 79.628535)], ignore_index=True)
print(kunden.sample(5, random_state=1).round(2))                                     # fünf Zeilen als Beispiel
print(kunden.groupby('gruppe')['gewinn'].agg(['size', 'mean', 'std']).round(1))    # Frage 1: 62 € (s 60) gegen 50 € (s 80)
```
