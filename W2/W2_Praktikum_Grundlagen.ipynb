## 3 · Vergleichen: True und False

**Zwei** Gleichheitszeichen vergleichen. Das Ergebnis ist `True` oder `False`. Andere Vergleiche: `<`, `>`, `<=`, `>=`,
`!=` (ungleich). Mit `and` und `or` verbindet man Bedingungen. Wichtig im Praktikum: Ein Vergleich mit einer ganzen
Spalte liefert je Zeile ein True oder False – und für Spalten heißt „und“ nicht `and`, sondern `&` (Abschnitt 7).

```python
verspaetung = 60
print(verspaetung > 30)            # True: 60 ist größer als 30
print(verspaetung == 15)           # False: zwei Gleichheitszeichen vergleichen
print(verspaetung > 30 and verspaetung < 100)
```

## 4 · Funktionen aufrufen

Eine Funktion ist ein Befehl mit runden Klammern: `round(3.14159, 2)` rundet auf 2 Stellen, `max(werte)` gibt den
größten Wert. Was in der Klammer steht, heißt **Argument**. Viele Funktionen „gehören“ zu einem Wert und werden mit
**Punkt** aufgerufen: `werte.count(15)` zählt, wie oft 15 in der Liste steht. Im Praktikum sieht das so aus:
`tage.mean()` – „der Mittelwert von tage“. **Punkt heißt: gehört dazu. Klammern heißen: ausführen.**

```python
print(round(3.14159, 2))           # Funktion mit zwei Argumenten in der Klammer
print(max(werte), min(werte))      # größter und kleinster Wert
print(werte.count(15))             # Punkt-Funktion: gehört zur Liste
```

## 7 · Zeilen filtern

Eine Bedingung auf einer Spalte gibt je Zeile True oder False. Steht diese Bedingung in eckigen Klammern hinter der
Tabelle, bleiben nur die Zeilen mit True. `(bedingung).mean()` ist der Anteil der True-Werte – im Praktikum ist das
der Anteil verspäteter Lieferungen. Zwei Bedingungen verbindet man bei Spalten mit `&` („und“) oder `|` („oder“),
jede in runden Klammern.

```python
ist_reich = kneipe['vermoegen'] > 50               # je Zeile True oder False
print(ist_reich.sum(), 'Gäste über 50 T€ · Anteil', ist_reich.mean())
mittelfeld = (kneipe['vermoegen'] > 20) & (kneipe['vermoegen'] < 100)    # zwei Bedingungen: & heißt „und“
print(mittelfeld.sum(), 'Gäste zwischen 20 und 100 T€')
kneipe[ist_reich]                                  # nur die Zeilen mit True
```

## 8 · Schleife und Dictionary

Eine **Schleife** führt dieselben Zeilen für jedes Element einer Liste aus: `for n in [3, 6, 9]:` – mit Doppelpunkt,
und darunter **eingerückt** (4 Leerzeichen) die Zeilen, die für jedes n laufen. Ein **Dictionary** speichert Werte
unter Namen, in geschweiften Klammern: `preise = {'Kaffee': 2.5}`. `preise['Kaffee']` holt den Wert,
`preise['Tee'] = 1.8` legt einen neuen an. Im Praktikum (Aufgabe 5) füllt eine Schleife ein Dictionary:
Stichprobengröße → Anteil.

```python
preise = {'Kaffee': 2.5, 'Tee': 1.8}            # Dictionary: Name → Wert
preise['Kakao'] = 3.0                           # neuen Eintrag anlegen
for getraenk in ['Kaffee', 'Tee', 'Kakao']:     # für jedes Element der Liste …
    print(getraenk, 'kostet', preise[getraenk], '€')   # … läuft diese eingerückte Zeile einmal
```

## 9 · Je Gruppe rechnen

`groupby` teilt eine Tabelle in Gruppen und rechnet je Gruppe eine Kennzahl: „gruppiere nach …, nimm die Spalte …,
Median je Gruppe“. Lesen Sie die Kette von links nach rechts, jeder Punkt ist ein Schritt. Im Praktikum sind die
Gruppen die Bundesstaaten.

```python
kneipe['beruf'] = ['lernt', 'lernt', 'arbeitet', 'arbeitet', 'arbeitet', 'arbeitet', 'arbeitet', 'arbeitet', 'Rente']
kneipe.groupby('beruf')['vermoegen'].median()      # je Gruppe der Median
```

```python
# TODO 9: je Gruppe Mittelwert und Anzahl auf einmal: statt .median() am Ende die Funktion .agg(['mean', 'size'])
je_gruppe = None

# --- Check ---
if je_gruppe is None:
    print('… noch offen')
else:
    print(je_gruppe)
    ok = hasattr(je_gruppe, 'loc') and abs(je_gruppe.loc['lernt', 'mean'] - 5) < 1e-9 and je_gruppe.loc['arbeitet', 'size'] == 6
    print('✓ richtig' if ok else '✗ prüfen: erwartet „lernt“ mit Mittelwert 5 und „arbeitet“ mit 6 Personen')
```
