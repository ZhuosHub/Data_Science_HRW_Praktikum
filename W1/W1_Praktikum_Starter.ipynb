## 2 · Fünf Fragen an Olist (ca. 30 Minuten)

Das sind die Schritte zwei und drei aus der Vorlesung: die Daten verstehen und aufbereiten. Jede Frage gibt die
SQL-Abfrage vor, Sie schreiben das Gegenstück in pandas. Die Tabelle `customers` heißt bei uns `kunden`.

| SQL | pandas | Bemerkung |
|---|---|---|
| `WHERE x = 'a'` | `t[t['x'] == 'a']` | innen entsteht eine True/False-Spalte, außen wird gefiltert |
| `JOIN t2 ON t.k = t2.k` | `t.merge(t2, on='k')` | Standard ist INNER JOIN |
| `GROUP BY g` + `COUNT(*)` | `t.groupby('g').size()` | |
| `GROUP BY g` + `AVG(x)` | `t.groupby('g')['x'].mean()` | ebenso `.sum()`, `.max()` … |
| `ORDER BY n DESC` | `.sort_values(ascending=False)` | |
| `LIMIT 5` | `.head(5)` | |
| `COUNT(*)` | `len(t)` | |
| `COUNT(DISTINCT x)` | `t['x'].nunique()` | |
